```python

import random

N = 4

def calculate_cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            same_row = state[i] == state[j]
            same_diagonal = abs(state[i] - state[j]) == abs(i - j)

            if same_row or same_diagonal:
                conflicts += 1

    return conflicts


def get_best_neighbor(state):
    best_state = None
    best_cost = calculate_cost(state)

    for col in range(N):
        for row in range(1, N + 1):
            if row == state[col]:
                continue

            neighbor = state.copy()
            neighbor[col] = row
            cost = calculate_cost(neighbor)

            if cost < best_cost:
                best_state = neighbor
                best_cost = cost

    return best_state, best_cost


def hill_climbing():
    restart = 0

    while True:
        restart += 1
        current = [random.randint(1, N) for _ in range(N)]

        print(f"\nRestart {restart}")
        print("Initial state:", current,
              "Cost:", calculate_cost(current))

        while True:
            current_cost = calculate_cost(current)

            if current_cost == 0:
                print("Goal state:", current)
                print("Final cost:", current_cost)
                return current

            neighbor, neighbor_cost = get_best_neighbor(current)

            if neighbor is None:
                print("Local minimum reached. Restarting...")
                break

            print("Neighbor:", neighbor,
                  "Cost:", neighbor_cost)

            current = neighbor


solution = hill_climbing()

print("\nFinal board:")
for row in range(1, N + 1):
    print(" ".join(
        "Q" if solution[col] == row else "."
        for col in range(N)
    ))
```

```python

from itertools import product

def implies(a, b):
    return (not a) or b


def is_kb_true(P, Q, R):
    sentence1 = implies(Q, P)
    sentence2 = implies(P, not Q)
    sentence3 = Q or R

    return sentence1 and sentence2 and sentence3


def entails(query):
    for P, Q, R in product([False, True], repeat=3):

        if is_kb_true(P, Q, R):
            if not query(P, Q, R):
                return False

    return True


print("P Q R | Q->P | P->not Q | Q or R | KB")
print("-" * 45)

for P, Q, R in product([True, False], repeat=3):
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R
    kb = s1 and s2 and s3

    def tf(value):
        return "T" if value else "F"

    print(
        tf(P), tf(Q), tf(R), "|",
        tf(s1), "  |", tf(s2), "       |",
        tf(s3), "     |", tf(kb)
    )

print("\nModels where KB is true:")

for P, Q, R in product([True, False], repeat=3):
    if is_kb_true(P, Q, R):
        print(f"P={P}, Q={Q}, R={R}")

queries = [
    ("R", lambda P, Q, R: R),
    ("R -> P", lambda P, Q, R: implies(R, P)),
    ("Q -> R", lambda P, Q, R: implies(Q, R))
]

print("\nEntailment results:")

for name, query in queries:
    result = entails(query)
    print(f"KB entails {name}: {result}")
```
